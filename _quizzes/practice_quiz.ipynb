**Question (C)** Plotting functions with Numpy

<font size = "3">

Using Numpy arrays, create a **line plot** of the function $f(g(t))$, where

$$ f(x) = x^2 - 5x + 2 $$

$$ g(t) = \sin(t + 3) $$

at a set of 75 evenly spaced points in the interval $2\pi \leq t \leq 4\pi$.

```python
# your code here
import numpy as np

def func_A(x: float) -> float:
    return x**2 - 5*x + 2
def func_B(t: float) -> float:
    return np.sin(t + 3)

x = np.linspace((2*np.pi), (4*np.pi), 75)
y = func_A(func_B(x))

plt.xlabel("x")
plt.ylabel("f(g(x))")
plt.title("f(g(x)) graphed over 2pi to 4pi")
plt.plot(x, y)
plt.show()
```

**Question (E)** Floating Point equality

<font size = "3">

The next code cell defines two variables (`x` and `y`) that in theory, should be equal. But when the cell is run, it prints out a message saying they are not the same. 

How would you choose a different comparison operation between `x` and `y` so that Python prints `True` instead?

```python
x = 0.1 * 7
y = 0.7

print("x and y are equal:", x == y)


# Your code here - choose a different comparison operator
print(abs(x-y))

epsilon = 1e-9

def float_p_check(x, y, error):
    if (abs(x-y) <= error):
        return True
    return False

print(f"x and y are floating point equal: {float_p_check(x,y,epsilon)} to the error term of {epsilon}")
```

**Question (F)** Testing membership in a list

<font size = "3">

The following code cell reads in a text file and converts it to a Python list. The list contains the declared majors of students in DATASCI 185, stored as strings.

Use an appropriate Boolean operation to create a variable with the value `True` if there is a student in the class majoring in Biology, and `False` otherwise. Print its value to the screen.

```python
# This code reads in a text file and converts it to a Python list
with open("quiz_data/majors.txt") as f:
    majors = [line.strip() for line in f]


# Your code here
def check_major(major, majors):
    return major in majors

print(check_major("Biology", majors))
```
