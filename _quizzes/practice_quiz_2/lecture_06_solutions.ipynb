```python
# create empty lists
good_movies = [] 
other_movies = []

cut_off = 7.0 # minimum rating for "good movie"

movie_titles = ["The Umbrellas of Cherbourg", "M3gan", "Psycho"]
movie_ratings = [7.8, 6.3, 8.5]

for index in range(len(movie_titles)):
    current_title = movie_titles[index]
    current_rating = movie_ratings[index]
    if current_rating >= cut_off:
        good_movies = good_movies + [current_title]
    else:
        other_movies = other_movies + [current_title]

print(good_movies)
print(other_movies)
```

#### **2)**. Using a `for` loop, compute the sum of all integers starting from 1 to the number stored in the variable $n$.

#### That is, compute the sum

$$ S_n = 1 + 2 + 3 + \cdots + n$$

<font size = "4">

**Solution**

This question appears on the practice quiz, and the most straightforward solution is given there. Here is another possible solution, where the sequence we loop over is 0, 1, 2, ..., n-1. So we need to add 1 to the loop variable to get the next term:

```python
n = 31 # run this cell with other positive integers (start smaller to check your answer)

Sn = 0
sequence = range(n) # This sequence is 0, 1, 2, ..., n-2, n-1 (the value of n is not included)

for loop_variable in sequence:
    term = loop_variable + 1
    Sn = Sn + term

print(Sn)
```

#### **3)**. Using a `for` loop, generate 3 scatter plots using the car data from "features.csv" in the "data" subfolder. The scatter plots should plot:

- #### displacement vs. mpg
- #### weight vs. mpg
- #### acceleration vs. mpg

```python
import pandas as pd
import matplotlib.pyplot as plt
carfeatures = pd.read_csv("data/features.csv")

# SOLUTION (leaving out axis labels and title since it wasn't in the problem description)

y_var = 'mpg'
x_variables = ['displacement', 'weight', 'acceleration']

for x_var in x_variables:
    plt.scatter(x = carfeatures[x_var], y = carfeatures[y_var])
    plt.show()
```
