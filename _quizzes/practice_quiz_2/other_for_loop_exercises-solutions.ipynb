**Q3** Compute sum of squares

<font size = "3">

- In the cell below, an integer `n` is randomly chosen satisfying $2 \leq n \leq 15$

- Using a "for" loop, compute the sum

$$ T_n = 1^2 + 2^2 + 3^2 + \cdots + n^2$$

- To check your work, I've added a variable called `expected_answer` that you can use to check your work. The sum computed in your for loop should match it.



````python
import random

n = random.randint(2, 15)
# n = 3     # feel free to make n smaller until you come up with your answer

# SOLUTION
Tn = 0
for k in range(1,n+1):
    term = k**2
    Tn = Tn + term



# post-loop code
print('The value of n is:', n)
expected_answer = int(n * (n + 1) * (2*n + 1) / 6)
print('The sum should have the value:', expected_answer)
print(Tn)
````

**Q4** Sum all positive elements

<font size = "3">

Below, we create a random `numpy` array with positive and negative numbers. Complete the "for" loop to compute **two** sums:

- `total_sum`: contains the sum of **all** elements in the array.

- `positive_sum`: contains the sum of all **positive** elements in the array.

**Note:** The keyword `pass` just means "do nothing" and it allows you to write an empty for loop without causing an error. Remove it and put your own code inside the for loop

````python
import numpy as np
N = 10
random_array = np.random.uniform(low = -5, high = 10, size = N)
total_sum = 0
positive_sum = 0

# SOLUTION
for val in random_array:
    total_sum = total_sum + val
    if val > 0:
        positive_sum = positive_sum + val


print(total_sum)
print(positive_sum)
````

**Q5** For loop with conditional statement

<font size = "3">

Suppose a cardiology team wants to flag patients who are at-risk of hypertension complications. They use the following rule:

- A patient is at risk if their systolic blood pressure is **greater than or equal to** 140 **and** they are **65 or older**

In the code block below, 3 lists are created containing information for 10 patients:

- `patient_id_lst` contains unique identification numbers for the 10 patients.

- `systolic_bp_lst` contains systolic blood pressure values for the 10 patients.

- `age_lst` contains the age (in years) for the 10 patients.

The ordering of the lists are consistent, so that patient "A1023" is 72 years old with a systolic blood pressure of 138. Patient "A1044" is 60 years old with a systolic blood pressure of 142, etc.

Using a **for** loop, append the patient IDs of all "at-risk" patients (those who meet the criteria listed above) to the initially empty list `at_risk`.

Then **print** `at_risk` to the screen.



````python
patient_id_lst   = ["A1023", "A1044", "B2331", "B2440", "C3001", "C3120", 
    "D4105", "D4119", "E5202", "E5307"]
systolic_bp_lst  = [138, 142, 152,  171, 155,  119, 144, 140, 132, 139]
age_lst          = [72,   60,   68,  65,  82,   34,  55,  70,  67, 75]

at_risk = []

bp_cut_off = 140
age_cut_off = 65

N = len(patient_id_lst)


# SOLUTION
for index in range(N):
    bp = systolic_bp_lst[index]
    id_no = patient_id_lst[index]
    age = age_lst[index]

    if (age >= age_cut_off) and (bp >= bp_cut_off):
        at_risk = at_risk + [id_no]

print(at_risk)
````

**Q7** Using enumerate

<font size = "3">

Using the `enumerate` function, write a for loop that prints out the following strings:

- "Movie 1 is The Wailing (2016)"
- "Movie 2 is Häxan (1922)"
- "Movie 3 is Gone Girl (2014)"
- "Movie 4 is Rosemary's Baby (1968)"
- "Movie 5 is Le Samouraï (1967)"

(Note that we are now printing that the first movie is "Movie 1", not "Movie 0")

<font size = "4">

**Solution 1**

````python
movie_titles = ["The Wailing", "Häxan", "Gone Girl", "Rosemary's Baby", "Le Samouraï"]
movie_years = ["2016", "1922", "2014", "1968", "1967"]

# SOLUTION 1

for index, title in enumerate(movie_titles):
    year = movie_years[index]
    print_string = "Movie " + str(index + 1) + " is " + title + " (" + year + ")"
    print(print_string)
````
