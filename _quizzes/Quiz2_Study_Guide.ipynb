# Part 1 — Objects, Types, and Variables

Almost everything in Python is an **object**, and every object has a **type** (a.k.a. class).
Common types: `int` (whole numbers), `float` (decimals), `str` (text), `bool` (`True`/`False`), `list`, `tuple`.

Use `type(obj)` to see an object's type, and `print(...)` to display things.


````python
x = 3
y = 3.25
s = "Hello"
flag = True

print(type(x))        # int
print(type(y))        # float
print(type(s))        # str
print(type(flag))     # bool
print(type(3.))       # 3. is still a float
````

### Variables are assigned with `=`

The **assignment operator** `=` is *not* an equals sign. Everything on the **right** is computed first, then
the name on the left is bound to it. Re-assignment is normal:

```python
z = 13
z = z + 1     # the new z uses the old z; result is 14
```

- Names may contain letters, digits, and `_`; they **cannot start with a digit**.
- Names are **case sensitive**: `first_name` and `First_name` are different.
- Prefer **descriptive** names.


````python
z = 13
print(z)
z = z + 1
print(z)
````

### `round()` — rounding numbers

`round(value, digits)` rounds to a given number of decimal places. Very common for money and for tidy output.


````python
record_price = 32.99
fulton_cost = record_price + record_price * 0.089
print(fulton_cost)                 # messy float
print(round(fulton_cost, 2))       # tidy to 2 decimals
````

# Part 2 — Numbers, Arithmetic, and the Float Trap

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `+` | add | `7 + 3` | `10` |
| `-` | subtract | `7 - 3` | `4` |
| `*` | multiply | `7 * 3` | `21` |
| `/` | divide (float) | `7 / 2` | `3.5` |

| `%` | remainder (modulo) | `7 % 2` | `1` |

````python
print(7 + 3, 7 - 3, 7 * 3, 7 / 2)
print(7 // 2, 7 % 2, 2 ** 3)
````

### ⚠️ Never test two floats for equality with `==`

Floating-point numbers are stored approximately. `0.1 + 0.1 + 0.1` is **not exactly** `0.3`.

Instead of `x == y`, check that they are *close*: `abs(x - y) < tol`, with a small tolerance like `1e-12`.


````python
x = 0.1 + 0.1 + 0.1
y = 0.3
print(x == y)                 # False  -- surprising but true!
print(abs(x - y) < 1e-12)     # True   -- the correct way
print(np.isclose(x, y))       # numpy also offers this (default tol ~1e-8)
````

# Part 3 — Strings

A string is a **sequence of characters**. Create with single or double quotes (don't mix them in one string).
Strings are **immutable** — you cannot change a character in place.

- Indexing: `s[0]` is the **first** character (0-based).
- Negative indexing: `s[-1]` is the **last** character.
- Slicing: `s[a:b]` gives characters from index `a` up to (not including) `b`.
- `len(s)` = number of characters.
- `+` **concatenates** two strings.
- `in` tests whether a substring appears inside a string.


````python
s = "Peter Sentz"
print(len(s))
print(s[0], s[6])          # 'P' , 'S'
print(s[0:5])              # 'Peter'
print(s[-1])               # 'z'
print("Peter" + " is here")
print("Peter" in s)        # True  -- substring test
print("peter" in s)        # False -- case sensitive
print(s.upper(), s.lower())
````

**Immutable:** this is an error, not allowed:

```python
s[0] = "p"      # TypeError: 'str' object does not support item assignment
```

**The `in` operator (important for Question C).** `keyword in text` is `True` when `keyword` appears anywhere in
`text`. Beware that `"Introduction"` is **not** found inside `"Intro to Python"` — `"Intro"` is a shorter word, so
it does not contain the full `"Introduction"`. Exact spelling matters.


````python
sentence = "The Federal Reserve makes forecasts about many economic outcomes"
print("economic" in sentence)   # True
print("bank" in sentence)       # False
print("Federal" in sentence)    # True  (capital F)
print("federal" in sentence)    # False (case sensitive)
print("Introduction" in "Intro to Python")   # False!
print("Introduction" in "Introduction to Python")  # True
````

# Part 4 — Lists (and Tuples)

A **list** stores multiple objects in one place. Created with square brackets `[...]`, items separated by commas.
Lists are **mutable** (you can change them).

- `lst[i]` accesses element `i` (0-based).
- `lst[i] = value` replaces an element.
- `lst.append(value)` adds to the end.
- `len(lst)` = number of elements.
- `+` concatenates two lists.
- `in` tests membership (exact match of an element).
- Lists may hold mixed types, and even other lists (nested).


````python
movies = ["Holes", "Almost Famous", "Psycho"]
print(movies)
print(len(movies))
print(movies[0], movies[-1])

movies[0] = "Bee Movie"      # mutate an element
movies.append("The Wailing") # add to end
print(movies)

print("Psycho" in movies)    # membership test
print(movies + ["Jaws"])     # concatenation (creates a NEW list)
````

### Tuples

A **tuple** is like a list but **immutable** — it cannot be changed after creation. Written with parentheses
(or even without them). Use a tuple when the values should never change.


````python
t = (5, 3, 2, 0)
print(t, type(t))
# t[0] = 9        # ERROR: tuples are immutable
t2 = 5, 3, 2, 0   # parentheses optional
print(t2, type(t2))
````

# Part 5 — Booleans, Comparison, and Logical Operators

`True` and `False` are the two Boolean **objects** (no quotes!). `"True"` (with quotes) is a **string**.

There are six comparison operators: `>`, `<`, `>=`, `<=`, `==` (equal), `!=` (not equal).
`==` compares; `=` assigns. Do not confuse them.

A handy fact: in numeric context, `True` behaves like `1` and `False` like `0`.


````python
print(type(True), type("True"))   # bool vs str
print(5 > 3, 5 >= 5, 3 != 3, 5 * 3 == 15)
print(True + 1, False - 1)        # 2 and -1
````

### Comparisons work on strings and lists too

- Two strings are `==` when every character matches exactly.
- Two lists are `==` when they have the same elements **in the same order**.

### Truthiness

A non-Boolean value used where a Boolean is expected is "forced": empty things
(`[]`, `""`, `0`, `None`) are **Falsy**; non-empty/zero things are **Truthy**.


````python
print("Hello World!" == "Hello World!")   # True
print("ab" == "ba")                        # False
print([1, 2, 3] == [1, 2, 3])              # True
print([1, 2, 3] == [2, 3, 1])              # False (order matters)

for val in [[], [1, 2], 0, 3, "", "hi"]:
    print(repr(val), "->", "Truthy" if val else "Falsy")
````

### Logical operators: `and`, `or`, `not`, `^`

- `A and B` → True only if **both** are true.
- `A or B` → True if **at least one** is true.
- `not A` → flips the value.
- `A ^ B` → **exclusive or**: True if **exactly one** is true.

**Precedence warning (from Lecture 5):** `and` binds tighter than `or`. Use parentheses to be unambiguous.

```
age >= 18 or has_exception and available     # means:  age >= 18 or (has_exception and available)
(age >= 18 or has_exception) and available   # usually what you actually mean!
```


````python
age = 21
has_exception = False
available_room = True

print(age >= 18 or has_exception and available_room)      # True (ambiguous)
print((age >= 18 or has_exception) and available_room)    # True (explicit)

print(True ^ False)   # True  (exactly one)
print(True ^ True)    # False (both)
print(not (5 > 3))    # False
````

# Part 6 — Control Flow: `if` / `elif` / `else`

Indented blocks run only when a condition holds. `if` tests a condition; `elif` ("else if") tests another only if
the earlier ones failed; `else` catches everything remaining.

```python
if condition_1:
    ...
elif condition_2:
    ...
else:
    ...
```

- The `:` starts an indented block (press Tab).
- Only the **first** matching branch runs.
- `elif` is a contraction of "else if" — Python has no `else if` keyword.


````python
score = 85
if score >= 90:
    letter = "A"
elif score >= 80:
    letter = "B"
elif score >= 70:
    letter = "C"
else:
    letter = "F"
print(letter)
````

### Piecewise-defined functions (Question B)

A **piecewise** definition gives different formulas on different intervals. Translate each case into a branch,
from lowest interval upward, using `<=` boundaries so the ranges don't overlap. Assign the result to the
requested variable (`y`) and print it.


````python
# y = 0 if x<=0 ; 0.5x if 0<x<=0.5 ; 0.5(1-x) if 0.5<x<=1 ; 0 if x>1
x = 0.3
if x <= 0:
    y = 0
elif x <= 0.5:
    y = 0.5 * x
elif x <= 1:
    y = 0.5 * (1 - x)
else:
    y = 0
print(y)
````

| Call | Values |

| `range(1, 7, 2)` | 1,3,5 |
| `range(5, 0, -1)` | 5,4,3,2,1 |

Convert to a list with `list(range(...))` to preview it.


````python
print(list(range(6)))
print(list(range(1, 6)))
print(list(range(1, 7, 2)))
print(list(range(5, 0, -1)))
````

# Part 8 — `for` Loops

A `for` loop repeats an indented block **once per element** of a sequence.

```python
for loop_variable in sequence:
    <body>          # runs once for each value; loop_variable holds the current value
```

The **loop variable** is just a name — call it whatever is descriptive.

### The "unrolled → loop" method (used all over your course)

Write the repeated lines **unrolled** first (each with a hard-coded value), then replace the changing value with
the loop variable and indent the block. This is exactly how Questions D and E work.


````python
# unrolled
cars = ["KIA", "Ferrari", "Ford", "Tesla"]
car_name = "KIA"
print("Dear customer, we are writing about your " + car_name + " vehicle.")
car_name = "Ferrari"
print("Dear customer, we are writing about your " + car_name + " vehicle.")
car_name = "Ford"
print("Dear customer, we are writing about your " + car_name + " vehicle.")
car_name = "Tesla"
print("Dear customer, we are writing about your " + car_name + " vehicle.")
````

````python
# same thing as a for loop
cars = ["KIA", "Ferrari", "Ford", "Tesla"]
for car_name in cars:
    print("Dear customer, we are writing about your " + car_name + " vehicle.")
````

### Getting the *position* too

Three common ways to know the index while looping:

1. A manual counter you increment inside the loop.
2. `range(len(seq))` and index in with `seq[i]`.
3. `enumerate(seq)` which hands you **both** index and value.


````python
cars = ["KIA", "Ferrari", "Ford", "Tesla"]

# 1) manual counter
count = 1
for car in cars:
    print("Dear " + car + " customer, your position is " + str(count) + " on the waitlist.")
    count = count + 1

# 2) range + index
N = len(cars)
for i in range(N):
    print(cars[i], "at index", i)

# 3) enumerate
for index, car in enumerate(cars):
    print(index, car)
````

### Accumulators (Question F pattern)

An **accumulator** is a variable you initialize *before* the loop and update *inside* it. The two most common:

- **Sum:** start at `0`, add each value.
- **Build a list:** start with `[]`, concatenate or `.append(...)`.

For "sum the integers 1..n", the loop variable is the number itself, so loop over a range that **includes n**.
Watch the boundary: `range(n)` stops at `n-1`; you want `range(1, n+1)` (or `range(n+1)`, since adding 0 is harmless).


````python
data = [1, 2, 13, 4, -2, 11, 2.2]
total = 0
for value in data:
    total = total + value     # or: total += value
print(total)

# sum of 1..n  -- note range must include n
n = 31
total = 0
for i in range(1, n + 1):
    total = total + i
print(total, "== expected", n * (n + 1) // 2)
````

### `break` and `continue`

- `break` stops the loop immediately.
- `continue` skips the rest of *this pass* and goes to the next element.

Handy for skipping bad data (e.g. `None` or text mixed into a numeric list).


````python
data = [1, 2, None, 4, "missing", 11, 2.2]
total = 0
for value in data:
    if (type(value) != int) and (type(value) != float):
        print("Warning: non-numerical element:", repr(value), "-- skipping")
        continue
    total = total + value
print("sum of numeric values:", total)
````

# Part 9 — NumPy Arrays (brief)

`numpy` arrays behave like lists but comparisons happen **element by element** (returning an array of Booleans).
You can loop over them just like lists. Useful for random numbers: `np.random.uniform(low, high, size)`.


````python
a = np.array([1, 2, 3])
b = np.array([1, 2, 4])
print(a == b)          # element-wise
print(a >= 2)          # element-wise
print(2 in a)          # membership

random_array = np.random.uniform(low=-5, high=10, size=5)
print(random_array)

# sum only the positive elements
pos_sum = 0
for value in random_array:
    if value <= 0:
        continue
    pos_sum = pos_sum + value
print("sum of positives:", round(pos_sum, 3))
````

# Part 10 — Pandas DataFrames

`pandas` is the course's table library. `pd.read_csv("path.csv")` loads a **DataFrame**. Key facts:

- `df.columns` → the column names (an index you can iterate).
- `df["mpg"]` → one column (a Series); `.mean()` averages it.
- `len(df)` → number of **rows**.  `len(df.columns)` → number of **columns**.
- `round(df[col].mean(), 2)` → tidy mean.

Below we build a small DataFrame inline so this notebook runs anywhere. (In the quiz you'd instead write
`df = pd.read_csv("files/features_clean.csv")`.)


````python
df = pd.DataFrame({
    "mpg":          [18.0, 15.0, 24.0, 30.0, 21.0],
    "cylinders":    [8, 8, 4, 4, 6],
    "displacement": [307.0, 350.0, 113.0, 98.0, 200.0],
    "horsepower":   [130.0, 165.0, 95.0, 80.0, 105.0],
    "weight":       [3504.0, 3693.0, 2372.0, 2100.0, 2900.0],
    "acceleration": [12.0, 11.5, 15.0, 16.5, 14.0],
})
print(type(df))
print(df.columns)
print("rows:", len(df), " cols:", len(df.columns))
print("mean mpg:", round(df["mpg"].mean(), 2))
````

### Question E pattern: mean of every column, with a loop

The unrolled version indexes `df.columns[0]`, `df.columns[1]`, ... . To generalize, loop over `df.columns`
directly, **or** use `range(len(df.columns))`. Both print the same strings.


````python
# direct iteration (clearest)
df = pd.DataFrame({
    "mpg": [18.0, 15.0, 24.0, 30.0],
    "weight": [3504.0, 3693.0, 2372.0, 2100.0],
    "acceleration": [12.0, 11.5, 15.0, 16.5],
})

for val in df.columns:
    mean_val = round(df[val].mean(), 2)
    print("Mean value of", val, ":", mean_val)
````

````python
# index version generalized with len()
df = pd.DataFrame({
    "mpg": [18.0, 15.0, 24.0, 30.0],
    "weight": [3504.0, 3693.0, 2372.0, 2100.0],
    "acceleration": [12.0, 11.5, 15.0, 16.5],
})

num_cols = len(df.columns)
for col in range(num_cols):
    val = df.columns[col]
    mean_val = round(df[val].mean(), 2)
    print("Mean value of", val, ":", mean_val)
````

# Part 11 — Matplotlib (scatter plots)

`matplotlib.pyplot` draws plots. The pattern for a scatter plot:

```python
plt.scatter(x_values, y_values)
plt.xlabel("...")
plt.ylabel("...")
```

Question D turns three repeated scatter plots into a loop.


````python
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 9, 16, 25])
plt.scatter(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("x vs. y")
plt.show()
````

````python
# plotting loop (Question D pattern)
df = pd.DataFrame({
    "mpg": [18.0, 15.0, 24.0, 30.0],
    "displacement": [307.0, 350.0, 113.0, 98.0],
    "weight": [3504.0, 3693.0, 2372.0, 2100.0],
    "acceleration": [12.0, 11.5, 15.0, 16.5],
})
y_var = "mpg"
plot_variables = ["displacement", "weight", "acceleration"]

for x_var in plot_variables:
    plt.scatter(df[x_var], df[y_var])
    plt.xlabel(x_var)
    plt.ylabel(y_var)
    plt.title(x_var + " vs. " + y_var)
    plt.show()
````

# Part 12 — Recipe Book (the patterns quizzes reuse)

| Goal | Recipe |
|------|--------|
| Repeat for each item in a list | `for item in my_list:` |
| Repeat with the index available | `for i in range(len(my_list)):` then `my_list[i]` |
| Index **and** value at once | `for i, item in enumerate(my_list):` |
| Sum values | `total = 0` before loop; `total += x` inside |
| Build a list of "good" items | `good = []` before; `good = good + [item]` inside |
| Skip bad data | `if bad: continue` |
| Stop early | `if done: break` |
| Count how many match | `count = 0` before; `count += 1` inside `if` |
| Mean of each column | `for val in df.columns: round(df[val].mean(), 2)` |
| Plot each variable vs. a target | loop the x-variables; scatter + labels + `show()` inside |
| Piecewise value | `if/elif/elif/else`, boundaries with `<=` |
| Test a substring/keyword | `keyword in text` |

### Converting "unrolled" code into a loop — step by step
1. Spot the line(s) that repeat.
2. Find the value that changes each repetition (index, column name, list item).
3. Make it the loop variable and loop over the collection of those values.
4. Indent the repeated block under the `for`.
5. Replace the hard-coded value with the loop variable everywhere in the block.
