#### **2)**. Using a `for` loop, compute the sum of all integers starting from 1 to the number stored in the variable $n$.

#### That is, compute the sum

$$ S_n = 1 + 2 + 3 + \cdots + n$$

<font size = "4">

**Solution**

This question appears on the practice quiz, and the most straightforward solution is given there. Here is another possible solution, where the sequence we loop over is 0, 1, 2, ..., n-1. So we need to add 1 to the loop variable to get the next term:

````python
n = 31 # run this cell with other positive integers (start smaller to check your answer)

Sn = 0
sequence = range(n) # This sequence is 0, 1, 2, ..., n-2, n-1 (the value of n is not included)

for loop_variable in sequence:
    term = loop_variable + 1
    Sn = Sn + term

print(Sn)
````

**5)**. With the random library, you can simulate a dice roll by randomly generating an integer between 1 and 6, using the line of code 

 ``` python
    roll = random.choice([1,2,3,4,5,6])
```
(you can choose a different variable name than "roll" of course).

- Let's simulate an experiment where you see how many dice rolls it takes to roll a 6. We'll do it with a "for" loop, but there's one problem: how many passes through the loop should we do?

- Theoretically, we could roll the die infinitely many times without seeing a 6. This almost surely won't happen however.

- We'll take a huge number and write a "for" loop that exits the loop when a 6 is rolled.

- Complete the for loop below so that once the first six is rolled, it prints "Rolled a six after {k} rolls" where k is the number of rolls it took **counting from 1**

````python
import random

huge_number = 1_000_000_000 # This is 1 billion. You can use underscores as a stand-in for commas for readability.

# SOLUTION
for i in range(huge_number):
    roll = random.choice([1, 2, 3, 4, 5, 6])
    if roll == 6:
        print("Rolled a six after " + str(i+1) + " rolls")
        break
````

**6)**. You can loop over `numpy` arrays just like lists (and ranges). Below, we create a random numpy array with positive and negative numbers. Using a `for` loop with the `continue` keyword, compute the sum of all the **positive** elements of the array.

````python
import numpy as np
N = 25
random_array = np.random.uniform(low = -5, high = 10, size = N)

# SOLUTION

pos_sum = 0

for val in random_array:
    if val <= 0:
        continue

    pos_sum = pos_sum + val

print(pos_sum)
````

#### **7)**. In 1734, Leonhard Euler discovered the following identity:

$$ \frac{\pi^2}{6} = \frac{1}{1^2} + \frac{1}{2^2} + \frac{1}{3^2} + \frac{1}{4^2} + \cdots $$

#### The sum on the right is an **infinite** series. Of course we cannot perform infinitely many tasks on a computer.

#### What we **can** do is write a for loop where we go through the loop a huge number of times (like exercise 5). We then simply terminate the summation once the newest term $\dfrac{1}{k^2}$ becomes smaller than a pre-specified tolerance. Compute a truncated approximation to $\dfrac{\pi^2}{6}$ using a `for` loop with the `break` keyword.

````python
tol = 1e-6 # currently set to 10^(-6). Choose other values

# SOLUTION
huge_number = 1_000_000_000

sum_approx = 0
for k in range(1, huge_number):
    new_term = 1/k**2
    sum_approx = sum_approx + new_term
    if new_term < tol:
        break

print("Approximate value of sum: ", sum_approx)

## BONUS: Approximate pi
pi_approx = np.sqrt(6 * sum_approx)
print('Corresponding approximation to pi is:', pi_approx)
````
